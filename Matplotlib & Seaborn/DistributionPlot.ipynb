```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
```

```python
df   # total_bil, tip -> quantitative
```

```python
df["time"].unique()   #qualitative
```

```python
df["size"].unique()  #qualitative
```

### histplot 


```python
sns.displot(df['total_bill'], bins=20, kde=True)  # bins are the container size we can imagine #kde- kernel density eastimator
```

```python
plt.subplot(1, 2, 1)
sns.histplot(df["total_bill"], kde=True)

plt.subplot(1, 2, 2)
sns.histplot(df["tip"], kde=True)
```

### jointplot - for comparing 2 data cols 

## scatter plot 


```python
df.head()
```

```python

sns.jointplot(x="total_bill", y="tip",data=df,  kind='scatter')

```

```python
df[df["total_bill"] == df["total_bill"].max()]
```

```python
sns.jointplot(x="total_bill", y="tip",data=df,  kind='hex')
```

```python
sns.jointplot(x="total_bill", y="tip",data=df,  kind='reg')
```

## pairplot   - multiple plot comarison (see x and y axis), auto select quantative cols

```python
sns.pairplot(df, hue='sex', palette='rainbow')  # hue='colName' is used to devide the data based on some value, for sex its male and female
```

### rugplot -- you can ignore

```python
sns.rugplot(df)
```
