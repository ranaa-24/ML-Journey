# Matrix Plots 
a color-coded visual representation of data arranged in a grid of rows and column

- heatmap
- clustermap
- pivot table heatmap

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flight = sns.load_dataset('flights')
flight.head()
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

## Heatmap
used to plot the corelations of some data columns 

### coreation - between diffrent cols

- -1 to 1
- postive means -> smoving in same dir ->ex: inc total_bill, incs tip
- negtive means -> moving oppsite direction -> ex: inc in smoking, dec lifespan



```python
# get the neumeric data col only
tips_corr = tips[['total_bill', 'tip', 'size']]
tips_corr
```

```python
tips_corr.corr()   # gives the corelation matrix
```

```python
sns.heatmap(tips_corr.corr())
```

```python
sns.heatmap(tips_corr.corr(), annot=True)
```

```python
## clustermap

sns.clustermap(tips_corr.corr())
```

## pivot table and heatmap

```python
flight.head()
```

```python
pvt_flight = flight.pivot_table(values='passengers', index='month', columns='year')
pvt_flight
```

```python
sns.heatmap(pvt_flight)
```
