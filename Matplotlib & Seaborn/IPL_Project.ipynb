```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()

# get the percentage of toss winner who is match winner also... so we need to get how may matches were there 
total_matches = df.shape[0] #no. of rows 
percentage = (count * 100) / total_matches
percentage.round(2)   # 48.6% chances of winning if we won a toss
```

## top 2 scorer  - who get the highest score

```python
df.head()
```

```python
## we can do aggregate on the top_scorer names with their score, and add the score
df.groupby('top_scorer')['highscore'].sum() # we aggregate by top_scorer all the cols are in same for you can imagine for a particular value 
# and we are only interested in the highscore and we sum them
```

### keep iin mind when there involve 2 cols always try to groupby them

```python
df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False)
```

```python
df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
```
