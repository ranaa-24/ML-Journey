```python
import numpy as np
```

```python
arr = np.array([2, 3, 4, 6, 7, 8, 9, 12, 13, 16, 17, 23, 27, 34, 38, 201])
```

```python
arr
```

```python
 # finding quateriles
Q1 = np.percentile(arr, 25);
Q2 = np.percentile(arr, 50);
Q3 = np.percentile(arr, 75);
```

```python
iqr = Q3-Q1
iqr
```

```python
# upper fence and lower fancce
uf = Q3 + 1.5 * iqr;
lf = Q1 - 1.5 * iqr;
```

```python
uf, lf
```

```python
# removing outliers
ls = [];
for i in arr:
    if i >= lf and i <= uf:
        ls.append(i)

ls = np.array(ls)
arr
```

```python
import seaborn as sns
```

```python
# with outliers
sns.boxplot(x=arr)
```

```python
# without outliers 
sns.boxplot(x=ls)
```
