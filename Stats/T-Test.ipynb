```python
import numpy as np
from scipy import stats
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
```

```python
n = len(sample);
s = np.std(sample);
sample_mean = np.mean(sample);
popu_mean = 170
alpha = 0.05
```

```python
t_test = (sample_mean - popu_mean) / (s/np.sqrt(n));
t_test
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t_test), df=n-1))
p_value
```

```python
if p_value < alpha:
    print("Reject H0");
else:
    print("Accept");
```
