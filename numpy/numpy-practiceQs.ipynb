### Easy way

```python
def validSuduku(arr):
    for i in range(0, 9, 3):
        for j in range(0, 9, 3):
            a3x3seg = arr[i:i+3, j:j+3]
            if np.sum(a3x3seg) != 45:
                return False

    return True


validSuduku(arr)
```
