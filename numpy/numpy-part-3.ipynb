### Deep and shalow copy

```python
# slcie always create a deep copy for both ls and ndarray

ls = [1,2,3,4,5]
ls_copy = ls[:3]
for x in range(len(ls_copy)):
    ls_copy[x] *= 10

print(ls_copy)
print(ls)
```

```python
# shallow copy
arr = np.arange(1, 4)
brr = arr
# brr = arr.copy()  #deep copy
brr[0] = 110000
arr
```
