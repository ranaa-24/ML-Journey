# slicing of matrix

```python
mat = np.arange(1, 31).reshape(5, 6)
print(mat)
```

```python
mat[1]
```

```python
mat[1, 2]   #as mat[1][2]
```

`mat[<rowSlice>, <colSlice>]`

```python
mat[0:2, 1:4]    #slice row 0&1 and col 1to4
```

```python
mat[2:, 3:]
```

```python
#col slice
mat[:, 3]
```

### Boolean indexing

```python
arr = np.arange(1, 11)
print(arr)
```

```python
even_idx = arr % 2 == 0
print(even_idx)
```

```python
arr[even_idx]
```
